```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
iris=load_iris()

```

```python
df=pd.DataFrame(data=iris.data,columns=iris.feature_names)
df.head()
```

```python
df['target']=iris.target
df.head()
```

```python
iris.target_names
```

```python
df[df.target==2].head()
```

```python
df['flower_name']=df.target.apply(lambda x:iris.target_names[x])
df.head()
```

```python
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
df0=df[df.target==0]
df1=df[df.target==1]
df2=df[df.target==2]
plt.scatter(df0['sepal length (cm)'],df0['sepal width (cm)'],color='green',marker='o',label='setosa')
plt.scatter(df1['sepal length (cm)'],df1['sepal width (cm)'],color='red',marker='o',label='versicolor')
plt.xlabel('sepal length (cm)')
plt.ylabel('sepal width (cm)')
plt.legend()
plt.show()
```

```python

plt.scatter(df0['petal length (cm)'],df0['petal width (cm)'],color='green',marker='o',label='setosa')
plt.scatter(df1['petal length (cm)'],df1['petal width (cm)'],color='red',marker='o',label='versicolor')
plt.xlabel('petal length (cm)')
plt.ylabel('petl width (cm)')
plt.legend()
plt.show()
```

```python
from sklearn.model_selection import train_test_split
x=df.drop(['target','flower_name'],axis='columns')
y=df.target
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2)

```

```python
from sklearn.svm import SVC
model=SVC()
model.fit(x_train,y_train)
```

```python
model.score(x_test,y_test)
```
